```python
#q2:
#Checking the divisibility of n by every integer from 2 up to n-1; if any divides evenly, n isn't prime, otherwise it is 
def is_PRime (n):
    if(n==0 or n==1):
        return False
    for i in range (2,n):
        if (n%i==0):
            return False

    return True

is_PRime(7) 
```

```python
#  Check missing values and fill numeric missing marks with the mean.
print(df.isna().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
#  Calculate the mean and standard deviation of the Age column.
print("Mean:", df["Age"].mean())
print("Standard Deviation:", df["Age"].std())
```

```python
#  Group passengers by class and find the passenger with the highest fare in each class.
top_passengers = df.loc[df.groupby("Pclass")["Fare"].idxmax()]

print(top_passengers[["Pclass", "Name", "Fare"]])
```

```python
# Group passengers by class and calculate the average fare for each class.
average_fare = df.groupby("Pclass")["Fare"].mean()

average_fare.plot(kind="bar")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.title("Average Fare per Passenger Class")
plt.show()

# The bar chart shows the average fare paid by passengers in each passenger class.
```
